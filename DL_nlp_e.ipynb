# basic cleaning

```python
text ='''🚀 Learn AI & ML in 2026 with us! Visit https://www.jeevithaaiacademy.com
 💻✨<div>Get 50% OFF now!!! 💥 #AI #ML @everyone</div>'''
```

```python
def clean_text(text):
  text = text.lower()
  # remove html tags
  text = re.sub(r'<.*?>','',text)
  # remove url
  text = re.sub(r'http\S+|www\S+','',text)
  # remove emojis
  text = re.sub(r'[^\x00-\x7F]+','',text)
  # remove special chars
  text = re.sub(r'[^a-z0-9]',' ',text)
  # remove white spaces
  text = re.sub(r'\s+',' ',text).strip()
  return text
```

```python
clean_text(text)
```

# Tokenization

```python
from nltk.tokenize import word_tokenize,sent_tokenize
# word tokenization
text = "NLP is very interesting to learn"
words = word_tokenize(text)
[x for x in words]
```

```python
# Sentence Tokenization
text = 'Python is easy. Sql is hard. I like NLP'
sent = sent_tokenize(text)
[x for x in sent]
```

# Stop words

```python
from nltk.corpus import stopwords
stop_words=set(stopwords.words('english'))
```

```python
text = "NLP is very interesting to learn"
words = word_tokenize(text)
[x for x in words if x not in stop_words]
```

**Pos Tagging**

```python
from nltk import pos_tag
text = 'John has started studying NLP and it seems very intersting'
words = word_tokenize(text)
pos_tag(words)
```

| Word        | Tag | Meaning                       |
| ----------- | --- | ----------------------------- |
| NLP         | NNP | Proper Noun (name of a field) |
| is          | VBZ | Verb (3rd person singular)    |
| very        | RB  | Adverb                        |
| interesting | JJ  | Adjective                     |
| and         | CC  | Conjunction                   |
| easy        | JJ  | Adjective                     |
| to          | TO  | "to" (infinitive marker)      |
| implement   | VB  | Base form Verb                |

**Named Entity recognition**

```python
from nltk import ne_chunk
text = 'John is working at Apple company in Bengaluru'
words=word_tokenize(text)
tag = pos_tag(words)
ner = ne_chunk(tag)
print(ner)
```

```python
import spacy
model = spacy.load('en_core_web_sm')
```

**Word Tokenization**

```python
text = 'I like NLP'
words = model(text)
[x.text for x in words]
```

**Sentence Tokenization**

```python
text = 'python is easy. Sql is hard'
sentence = model(text)
[x.text for x in sentence.sents]
```

**Stop words**


```python
text = 'python is easy Sql is hard'
words = model(text)
[x.text for x in words if not x.is_stop]
```

# lemmatization

```python
text = 'playing played plays studies studying better'
words = model(text)
[x.lemma_ for x in words]
```

# Pos tag

```python
text = 'John has started studying NLP and it seems very intersting'
words = model(text)
{x:x.pos_ for x in words if not x.is_stop}
```

# NER

```python
text = 'John is working at Apple company in Bengaluru'
words = model(text)
{x:x.label_ for x in words.ents}
```
