**Encoding**

```python
# Gender - One hot encoding
# Married	,Self_Employed # Loan_Status (Approved): Label ENcoder
# Education - order matters
# Property_Area - target encoding
```

```python
data = pd.get_dummies(data,columns=['Gender'],dtype='int')
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
data.loc[:,'married']= le.fit_transform(data.loc[:,'Married'])
data.loc[:,'self_Employed']= le.fit_transform(data.loc[:,'Self_Employed'])
data.loc[:,'loan_status']=le.fit_transform(data.loc[:,'Loan_Status (Approved)'])
```

```python
# Education
data.loc[:,'education']=data.loc[:,'Education'].map({'Graduate':1,'Not Graduate':0})
```

```python
# Property_Area
plt.figure(figsize=(5,2))
sns.countplot(data,x='Property_Area',hue='loan_status')
plt.show()
```

```python
data.loc[:,'property_Area'] = data.loc[:,'Property_Area'].map(
    {'Semiurban':2,'Urban':1,'Rural':0})
```

```python
# replace 3+ with 3
data.loc[:,'dependents']= data.loc[:,'Dependents'].replace({'3+':3}).astype('int64')
```

# Scaling

```python
#	ApplicantIncome	CoapplicantIncome	LoanAmount
# standard scaler
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
data[['ApplicantIncome']]= scaler.fit_transform(data[['ApplicantIncome']])
data[['CoapplicantIncome']]= scaler.fit_transform(data[['CoapplicantIncome']])
data[['LoanAmount']]= scaler.fit_transform(data[['LoanAmount']])
```

**Split data into x and y**





```python
x = data.drop(columns='loan_status')
y = data.loc[:,'loan_status']
```

**Split data for training and testing**

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```
